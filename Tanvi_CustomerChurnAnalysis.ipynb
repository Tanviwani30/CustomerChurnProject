```python
# ============================================
# CUSTOMER CHURN ANALYSIS & PREDICTION
# Step 1 to Step 5 - Initial Setup
# ============================================

# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")

# Step 2: Load Dataset
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)

# Step 3: Display First 5 Rows
print("\nFirst 5 Rows:")
display(df.head())

# Step 4: Dataset Information
print("\nDataset Information:")
df.info()

# Step 5: Basic Statistics
print("\nBasic Statistics:")
display(df.describe())

# Check Churn Count
print("\nChurn Count:")
print(df["Churn"].value_counts())

# Check Churn Percentage
print("\nChurn Percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)

# ============================================
# STEP 6: DATA CLEANING
# ============================================

# 1. Check missing values
print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# 2. Check duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# 3. Convert TotalCharges to numeric
# Some values may be blank, so errors='coerce' converts them to NaN
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# 4. Check missing values again
print("\nMissing Values After Converting TotalCharges:")
print(df.isnull().sum())

# 5. Fill missing TotalCharges using median
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())

# 6. Remove duplicate rows if any
df = df.drop_duplicates()

# 7. Final dataset information
print("\nDataset After Cleaning:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nRemaining Missing Values:")
print(df.isnull().sum().sum())

print("\nData Cleaning Completed Successfully!")

# ============================================
# STEP 7: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================

# 1. Churn Distribution
plt.figure(figsize=(7, 5))
sns.countplot(x="Churn", data=df)
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()


# 2. Churn Percentage
churn_percentage = df["Churn"].value_counts(normalize=True) * 100

print("Churn Percentage:")
print(churn_percentage.round(2))


# 3. Gender vs Churn
plt.figure(figsize=(7, 5))
sns.countplot(x="gender", hue="Churn", data=df)
plt.title("Gender vs Customer Churn")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()


# 4. Contract Type vs Churn
plt.figure(figsize=(8, 5))
sns.countplot(x="Contract", hue="Churn", data=df)
plt.title("Contract Type vs Customer Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.show()


# 5. Internet Service vs Churn
plt.figure(figsize=(8, 5))
sns.countplot(x="InternetService", hue="Churn", data=df)
plt.title("Internet Service vs Customer Churn")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")
plt.show()


# 6. Monthly Charges vs Churn
plt.figure(figsize=(8, 5))
sns.boxplot(x="Churn", y="MonthlyCharges", data=df)
plt.title("Monthly Charges vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()


# 7. Tenure vs Churn
plt.figure(figsize=(8, 5))
sns.boxplot(x="Churn", y="tenure", data=df)
plt.title("Tenure vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()


print("\nEDA Completed Successfully!")

# ============================================
# STEP 8: FEATURE SELECTION & DATA PREPROCESSING
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 1. Remove customerID because it does not help in prediction
df_model = df.drop("customerID", axis=1).copy()

# 2. Convert Churn into numerical values
# No = 0, Yes = 1
label_encoder = LabelEncoder()
df_model["Churn"] = label_encoder.fit_transform(df_model["Churn"])

# 3. Separate Features (X) and Target (y)
X = df_model.drop("Churn", axis=1)
y = df_model["Churn"]

print("Features and target separated successfully!")
print("Features shape:", X.shape)
print("Target shape:", y.shape)

# 4. Convert categorical columns into numerical columns
X = pd.get_dummies(X, drop_first=True)

print("\nData after encoding:")
print("Number of features:", X.shape[1])

# 5. Convert all values to numeric
X = X.astype(float)

# 6. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nData Split Completed!")
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nPreprocessing Completed Successfully!")

# ============================================
# STEP 9: LOGISTIC REGRESSION WITH SCALING
# ============================================

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# 1. Create StandardScaler
scaler = StandardScaler()

# 2. Scale training and testing data
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully!")

# 3. Create Logistic Regression model
model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

# 4. Train the model
model.fit(X_train_scaled, y_train)

print("Model training completed successfully!")

# 5. Make predictions
y_pred = model.predict(X_test_scaled)

# 6. Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Accuracy:", round(accuracy * 100, 2), "%")

# ============================================
# STEP 10: MODEL EVALUATION
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# 1. Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

# 2. Display metrics
print("MODEL EVALUATION RESULTS")
print("=" * 40)

print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1-Score  : {f1 * 100:.2f}%")

# 3. Detailed Classification Report
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["No Churn", "Churn"]
))

# 4. Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Churn", "Churn"],
    yticklabels=["No Churn", "Churn"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

print("\nModel Evaluation Completed Successfully!")

# ============================================
# STEP 11: FEATURE IMPORTANCE / CHURN FACTORS
# ============================================

# Get feature names
feature_names = X.columns

# Get Logistic Regression coefficients
coefficients = model.coef_[0]

# Create a DataFrame
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

# Calculate absolute importance
feature_importance["Importance"] = feature_importance["Coefficient"].abs()

# Sort by importance
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

# Display top 15 important features
print("Top 15 Features Influencing Churn Prediction:")
display(feature_importance.head(15))

# Plot Top 15 Features
top_features = feature_importance.head(15).sort_values(
    by="Importance"
)

plt.figure(figsize=(10, 7))

sns.barplot(
    x="Importance",
    y="Feature",
    data=top_features
)

plt.title("Top 15 Features Influencing Customer Churn")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()

print("\nFeature Importance Analysis Completed Successfully!")

# ============================================
# STEP 12: NEW CUSTOMER CHURN PREDICTION
# ============================================

# Take one customer from the test dataset
sample_customer = X_test.iloc[[0]]

# Scale the customer data
sample_customer_scaled = scaler.transform(sample_customer)

# Make prediction
prediction = model.predict(sample_customer_scaled)[0]

# Get prediction probability
probability = model.predict_proba(sample_customer_scaled)[0]

# Display result
print("CUSTOMER CHURN PREDICTION")
print("=" * 40)

if prediction == 1:
    print("Prediction: CUSTOMER MAY CHURN")
else:
    print("Prediction: CUSTOMER MAY NOT CHURN")

print(f"\nProbability of No Churn: {probability[0] * 100:.2f}%")
print(f"Probability of Churn: {probability[1] * 100:.2f}%")

# ============================================
# STEP 13: MODEL RESULTS VISUALIZATION
# ============================================

# Count actual and predicted values
actual_counts = pd.Series(y_test).value_counts().sort_index()
predicted_counts = pd.Series(y_pred).value_counts().sort_index()

# Create comparison DataFrame
comparison = pd.DataFrame({
    "Actual": actual_counts,
    "Predicted": predicted_counts
})

comparison.index = ["No Churn", "Churn"]

print("Actual vs Predicted Customer Counts:")
display(comparison)

# Plot comparison
comparison.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Actual vs Predicted Customer Churn")
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.legend(title="Result")
plt.show()

# Display final model metrics
print("\nFINAL MODEL METRICS")
print("=" * 40)
print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1-Score  : {f1 * 100:.2f}%")

# ============================================
# STEP 14: FINAL INSIGHTS / FINDINGS
# ============================================

print("=" * 60)
print("FINAL DATASET INSIGHTS")
print("=" * 60)

# 1. Overall churn rate
churn_rate = (df["Churn"].value_counts(normalize=True)["Yes"]) * 100
print(f"\n1. Overall Customer Churn Rate: {churn_rate:.2f}%")

# 2. Average monthly charges by churn
monthly_charges = df.groupby("Churn")["MonthlyCharges"].mean()

print("\n2. Average Monthly Charges:")
print(monthly_charges.round(2))

# 3. Average tenure by churn
tenure_analysis = df.groupby("Churn")["tenure"].mean()

print("\n3. Average Tenure:")
print(tenure_analysis.round(2))

# 4. Contract type churn rate
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print("\n4. Churn Percentage by Contract Type:")
display(contract_churn.round(2))

# 5. Internet service churn rate
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

print("\n5. Churn Percentage by Internet Service:")
display(internet_churn.round(2))

# 6. Senior citizen churn rate
senior_churn = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

print("\n6. Churn Percentage by Senior Citizen Status:")
display(senior_churn.round(2))

# 7. Highest churn contract
highest_contract = contract_churn["Yes"].idxmax()
highest_contract_rate = contract_churn["Yes"].max()

print(
    f"\n7. Contract with Highest Churn Rate: "
    f"{highest_contract} ({highest_contract_rate:.2f}%)"
)

# 8. Highest churn internet service
highest_internet = internet_churn["Yes"].idxmax()
highest_internet_rate = internet_churn["Yes"].max()

print(
    f"8. Internet Service with Highest Churn Rate: "
    f"{highest_internet} ({highest_internet_rate:.2f}%)"
)

print("\n" + "=" * 60)
print("INSIGHTS ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 60)

# ============================================
# STEP 15: PROJECT CONCLUSION
# ============================================

print("=" * 60)
print("PROJECT CONCLUSION")
print("=" * 60)

print("""
This project analyzed customer churn using data analytics
and machine learning techniques.

The dataset was cleaned and explored to understand customer
characteristics and their relationship with churn.

Exploratory Data Analysis was performed using different
visualizations, including churn distribution, contract type,
internet service, monthly charges, and customer tenure.

A Logistic Regression model was developed to predict whether
a customer is likely to churn.

The model was evaluated using Accuracy, Precision, Recall,
F1-Score, Classification Report, and Confusion Matrix.

The analysis provides useful insights into customer behavior
and can help businesses identify customers who may be at risk
of leaving their service.

Overall, this project demonstrates how Data Analytics and
Machine Learning can be combined to analyze customer churn
and support data-driven decision making.
""")

print("=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
